## Tratamento de dados - Dataset B


#### Remover 1 duplicado

Primeiramente, precisamos importar as bibliotecas que utilizaremos para manipular os dados.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

Agora iremos carregar o dataset b

```python
dataset_b = pd.read_csv("../dados/brutos/dataset_B.csv")
```

Execução do código que apaga o valor duplicado


```python
df = pd.DataFrame(dataset_b)

df = df.drop_duplicates()

#print(df.duplicated().sum())
```

OBS: possívelmente podem exitir bairros escritos como SP. o que fazer com eles?

#### Extrair `bairro` da coluna `endereço_casa`

```python
# carregamos o datasetb

# pegamos os valores da coluna endereço_casa
s = df['endereço_casa']

# retirei o traço e pedi para o pandas dividir cada string em colunas
criar_indices = s.str.split("-", expand=True)

#retornei o índice em que a coluna do bairro estava (no caso, coluna1)
extrair_bairro = criar_indices[1]

# retirei a virgula que separava o bairro do estado
bairro = extrair_bairro.str.split(", ", expand=True)

#retornei a coluna que estava o bairro, que no caso era a 0
df["endereço_casa"] = bairro[0]
```

#### Converter `preco_casa` de texto para número

```python
s = df['preco_casa']

retirar_simbolo = s.str.replace({"R$ ": ""})

retirar_virgula = retirar_simbolo.str.replace({".":""})

retirar_caracteres_estranhos = retirar_virgula.str.replace({" /mês": ""})

conversao = retirar_caracteres_estranhos.astype("float")

df['preco_casa'] = conversao

```

#### Seleção das colunas


```python
df_final = df[["endereço_casa", "preco_casa", "metros_casa", "quartos_casa", "banheiros_casa", "vagas_casa"]]
```

#### Salvar no arquivo csv

```python
df_final.to_csv("../dados/tratados/dataset_treino.csv", index=False)
```
